# Business Understanding

## Contexto

Uma empresa do setor supermercadista pretende ofertar cartões de crédito aos seus clientes em parceria com uma instituição financeira.

Para reduzir o risco de concessão de crédito, foi disponibilizada uma base histórica contendo informações dos clientes e seu comportamento de pagamento, permitindo o desenvolvimento de modelos capazes de prever a probabilidade de inadimplência.

Além da base histórica ("Base de Treino"), foi disponibilizada uma segunda base ("Base sem performance"), contendo clientes sem histórico conhecido de inadimplência, que será utilizada para aplicação do modelo desenvolvido.

# Objetivo do Projeto

Desenvolver um modelo de Machine Learning capaz de estimar a probabilidade de inadimplência de clientes, auxiliando o processo de concessão de crédito através da identificação de clientes com maior ou menor risco.

O modelo deverá ser capaz de classificar novos clientes e gerar uma probabilidade de inadimplência que poderá ser utilizada como apoio à decisão na oferta de cartões pré-aprovados.

# Problema de Negócio

A concessão de crédito para clientes com alto risco de inadimplência pode gerar perdas financeiras significativas para a instituição.

Por outro lado, negar crédito para clientes com baixo risco reduz oportunidades de negócio e impacta negativamente os resultados da empresa.

O desafio consiste em construir um modelo capaz de equilibrar risco e oportunidade, permitindo decisões mais assertivas durante o processo de aprovação de crédito.

# Objetivo Analítico

O problema será tratado como um problema de Classificação Binária Supervisionada.

Variável resposta:

- **0** → Cliente adimplente
- **1** → Cliente inadimplente

Serão avaliados diferentes algoritmos de Machine Learning para identificar aquele com melhor capacidade de discriminar clientes adimplentes e inadimplentes.

# Critérios de Sucesso

O desempenho dos modelos será comparado utilizando métricas amplamente empregadas em problemas de risco de crédito.

As principais métricas avaliadas serão:

- ROC AUC
- KS (Kolmogorov-Smirnov)
- Gini
- Accuracy
- Precision
- Recall
- F1-Score

Além das métricas quantitativas, também serão consideradas a interpretabilidade do modelo e a importância das variáveis para a tomada de decisão.

# Hipóteses Iniciais

Antes da exploração dos dados, algumas hipóteses são consideradas:

- Existem variáveis mais relevantes que outras para explicar a inadimplência.
- Algumas variáveis podem apresentar baixa capacidade preditiva e poderão ser removidas.
- O conjunto de dados pode apresentar valores ausentes, outliers ou desbalanceamento entre as classes.
- Modelos baseados em árvores de decisão, como XGBoost e LightGBM, podem apresentar desempenho superior à Regressão Logística.
- Será necessário realizar etapas de tratamento e preparação dos dados antes da modelagem.

# Metodologia

O desenvolvimento seguirá uma abordagem estruturada baseada no processo CRISP-DM.

As etapas do projeto serão:

1. Business Understanding
2. Data Understanding (EDA)
3. Data Preparation (Feature Engineering)
4. Modeling
5. Evaluation
6. Prediction
7. Conclusões e Recomendações

# Algoritmos Avaliados

Serão comparados três algoritmos amplamente utilizados em problemas de classificação e modelagem de risco de crédito:

- Regressão Logística
- XGBoost
- LightGBM

A escolha do modelo final será baseada nas métricas de desempenho e na capacidade de generalização observada durante a validação.

# Resultado Esperado

Ao final do projeto espera-se obter um modelo capaz de estimar a probabilidade de inadimplência para novos clientes, permitindo apoiar a decisão de concessão de crédito.

O modelo será aplicado à "Base sem performance", produzindo uma probabilidade de inadimplência para cada cliente, que poderá ser utilizada como apoio à estratégia de oferta de cartões pré-aprovados.

# Entregáveis

Ao final deste projeto serão entregues:

- Análise Exploratória dos Dados (EDA);
- Tratamento e preparação dos dados;
- Seleção das variáveis explicativas;
- Comparação entre diferentes algoritmos de Machine Learning;
- Avaliação utilizando métricas de classificação;
- Interpretação dos resultados do modelo;
- Predição da probabilidade de inadimplência para a Base sem performance;
- Recomendação sobre a utilização do modelo no processo de concessão de crédito.
