## Considerações sobre o Modeling

Foram testados três algoritmos de classificação com diferentes características: Regressão Logística, XGBoost e LightGBM.

A avaliação dos modelos foi realizada utilizando métricas como **ROC AUC, Gini, KS, Precision, Recall e F1-Score**, permitindo analisar tanto a capacidade discriminativa dos modelos quanto seu desempenho na identificação dos clientes inadimplentes.

O **XGBoost apresentou o maior ROC AUC e Gini**, indicando uma maior capacidade discriminativa global. Entretanto, o **LightGBM apresentou o maior KS e F1-Score**, além de um Recall significativamente superior ao XGBoost.

Considerando o contexto do problema, no qual a identificação de clientes com maior risco de inadimplência é relevante para a decisão de concessão de crédito, o **LightGBM foi selecionado como modelo final**.

A seleção das variáveis explicativas foi realizada de forma criteriosa, considerando os resultados da análise exploratória, correlações, Feature Importance, Permutation Importance e os coeficientes da Regressão Logística. Após a redução das variáveis, o desempenho do modelo foi novamente avaliado para verificar se a simplificação mantinha uma capacidade preditiva adequada.

O modelo final será utilizado para estimar a **probabilidade de inadimplência dos clientes da base sem performance**, permitindo que o resultado seja utilizado como apoio à definição de políticas de concessão de crédito.
