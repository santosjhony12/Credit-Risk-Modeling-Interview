# Resumo do Feature Engineering

Nesta etapa foram realizados os procedimentos necessários para preparar a base de dados para a modelagem.

## Tratamentos realizados

- Remoção da coluna identificadora (ID), por não possuir capacidade preditiva.
- Remoção dos registros sem informação da variável resposta.
- Tratamento dos valores ausentes nas variáveis numéricas utilizando a mediana.
- Tratamento dos valores ausentes nas variáveis categóricas utilizando a categoria mais frequente.
- Codificação das variáveis categóricas por meio de One-Hot Encoding.
- Verificação da existência de variáveis constantes.
- Avaliação da correlação entre as variáveis para identificação de possíveis redundâncias.
- Exportação da base tratada para utilização na etapa de modelagem.

## Resultado

Ao final desta etapa foi obtida uma base limpa, estruturada e preparada para o treinamento dos modelos de Machine Learning.
