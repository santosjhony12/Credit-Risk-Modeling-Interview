# Resumo Técnico da Análise Exploratória dos Dados (EDA)

A base de treinamento é composta por **35.572 registros** e **21 variáveis**, sendo uma variável resposta (`PERFORMANCE (var resposta)`), um identificador (`ID`) e 19 variáveis explicativas anônimas (`V1` a `V19`).

## Qualidade dos Dados

A análise inicial identificou a presença de valores ausentes em diversas variáveis. O maior percentual de ausência foi observado na variável **V8**, com aproximadamente **60,24%** de valores nulos. A própria variável resposta apresenta aproximadamente **47,30%** de valores ausentes, o que é esperado, uma vez que a base disponibilizada contém tanto registros utilizados para treinamento quanto registros sem informação de performance.

Além dessas variáveis, outras também apresentam valores ausentes, porém em menor proporção, como V9, V2, V14 e V3.

Não foram identificados valores ausentes nas colunas **ID** e **V19**.

## Balanceamento da Variável Resposta

Considerando apenas os registros com informação de performance, observa-se que a base apresenta um leve desbalanceamento entre as classes.

- Clientes adimplentes (0): **80,05%**
- Clientes inadimplentes (1): **19,95%**

Embora exista predominância de clientes adimplentes, o nível de desbalanceamento não é considerado extremo para problemas de risco de crédito, permitindo inicialmente a utilização de algoritmos tradicionais de classificação. Ainda assim, durante a etapa de modelagem será avaliada a necessidade de técnicas para tratamento do desbalanceamento.

## Correlação das Variáveis

A análise de correlação mostrou que nenhuma variável apresenta correlação linear muito elevada com a variável resposta.

As maiores correlações observadas foram:

| Variável | Correlação |
|----------|-----------:|
| V10 | 0,180 |
| V3 | 0,065 |

As demais variáveis apresentaram correlações negativas de baixa magnitude, indicando que, individualmente, nenhuma variável explica completamente a inadimplência. Esse comportamento é esperado em problemas de crédito, nos quais o risco normalmente resulta da combinação de diversos fatores.

## Análise das Variáveis Categóricas

As variáveis categóricas apresentaram diferenças relevantes na distribuição da inadimplência.

Na variável **V12**, clientes classificados como **Inativo** apresentaram aproximadamente **29,18%** de inadimplência, enquanto clientes classificados como **Fiel** apresentaram apenas **12,42%**.

Comportamento semelhante foi observado na variável **V13**, na qual a categoria **Inativo** também apresentou maior percentual de inadimplentes (**28,88%**) em comparação com a categoria **Fiel** (**13,53%**).

Na variável **V17**, observou-se uma diferença ainda mais expressiva. Clientes classificados como **Premium** apresentaram aproximadamente **10,14%** de inadimplência, enquanto clientes classificados como **Inativo** atingiram cerca de **29,18%**.

Esses resultados indicam que essas variáveis possuem potencial capacidade discriminatória e deverão ser avaliadas durante a etapa de seleção de atributos.

## Considerações Finais

A análise exploratória permitiu identificar alguns pontos importantes para as próximas etapas do projeto:

- Existem variáveis com elevado percentual de valores ausentes, sendo necessário definir uma estratégia adequada de tratamento antes da modelagem.
- A variável resposta apresenta leve desbalanceamento, característica comum em problemas de concessão de crédito.
- Não foram observadas correlações lineares muito fortes entre as variáveis e a inadimplência, sugerindo que modelos capazes de capturar relações não lineares podem apresentar melhor desempenho.
- Algumas variáveis categóricas demonstraram boa capacidade de separar clientes adimplentes e inadimplentes, indicando potencial relevância para o modelo preditivo.
- Na próxima etapa serão realizados o tratamento dos valores ausentes, codificação das variáveis categóricas, seleção das variáveis explicativas e preparação dos dados para treinamento dos modelos de Machine Learning.
